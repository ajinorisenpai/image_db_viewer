```python
def insert_image(path, description, detail):
    try:
        sql_cursor.execute("""INSERT INTO images(path, description, detail) VALUES(?, ?, ?)
        ON CONFLICT(path) DO UPDATE SET
            description = excluded.description,
            detail = excluded.detail
        """,(path, description, detail))
        sql_conn.commit()
    except sqlite3.Error as e:
        print(f"エラー: {e}")
```

```python
import numpy as np
from math import sqrt, ceil
from PIL import Image, ImageDraw, ImageFont

def validate_input(array):
    """入力データの検証"""
    if len(array) == 0:
        raise ValueError("画像が提供されていません")
    return array[0].shape

def get_optimal_grid_size(num_images, max_rows, max_cols):
    """最適なグリッドサイズを計算"""
    optimal_cols = min(max_cols, ceil(sqrt(num_images)))
    optimal_rows = min(max_rows, ceil(num_images / optimal_cols))
    optimal_cols = min(max_cols, ceil(num_images / optimal_rows))
    return optimal_rows, optimal_cols

def load_font(font_size):
    """フォントの読み込み"""
    try:
        return ImageFont.truetype("arial.ttf", font_size)
    except IOError:
        try:
            return ImageFont.truetype("DejaVuSans.ttf", font_size)
        except IOError:
            return ImageFont.load_default()

def resize_image(img, target_size):
    """画像のリサイズ処理"""
    pil_img = Image.fromarray(img)
    return pil_img.resize(target_size, Image.LANCZOS)

def add_number_to_image(img, number, font, font_color):
    """画像に番号を追加"""
    pil_img = Image.fromarray(img)
    draw = ImageDraw.Draw(pil_img)

    number_text = str(number)
    text_position = (10, 10)
    left, top, right, bottom = draw.textbbox(text_position, number_text, font=font)

    # 背景の四角形を描画
    draw.rectangle(
        [left-5, top-5, right+5, bottom+5],
        fill=(0, 0, 0)
    )

    # テキストを描画
    draw.text(
        text_position,
        number_text,
        font=font,
        fill=font_color
    )

    return np.array(pil_img)

def create_blank_image(shape, channels=3):
    """空白画像の生成"""
    return np.ones((*shape, channels), dtype=np.uint8) * 255

def build_image_grid(processed_images, grid_size, target_size):
    """グリッド画像の構築"""
    grid_rows, grid_cols = grid_size
    target_height, target_width = target_size

    # グリッド画像の初期化
    grid_image = create_blank_image(
        (grid_rows * target_height, grid_cols * target_width)
    )

    # 画像をグリッドに配置
    for i, img in enumerate(processed_images):
        row = i // grid_cols
        col = i % grid_cols

        y_start = row * target_height
        y_end = y_start + target_height
        x_start = col * target_width
        x_end = x_start + target_width

        grid_image[y_start:y_end, x_start:x_end] = img

    return grid_image

def image_grid(array, max_nrows=10, max_ncols=10, add_numbers=True, font_size=20, font_color=(255,255,255)):
    """
    画像の配列からグリッドを作成するメイン関数
    異なるサイズの画像にも対応
    """
    # 入力検証と基本情報取得
    target_height, target_width, channels = validate_input(array)
    num_images = len(array)
    target_size = (target_width, target_height)

    # グリッドサイズ計算
    grid_rows, grid_cols = get_optimal_grid_size(num_images, max_nrows, max_ncols)
    required_images = grid_rows * grid_cols

    # フォント読み込み
    font = load_font(font_size)

    # 画像処理パイプライン
    processed_images = []
    for idx, img in enumerate(array, 1):
        # リサイズ処理
        if img.shape[:2] != (target_height, target_width):
            pil_img = resize_image(img, target_size)
            img = np.array(pil_img)

        # 番号追加処理
        if add_numbers:
            img = add_number_to_image(img, idx, font, font_color)

        processed_images.append(img)

    # 不足分を空白画像で補完
    blank_image = create_blank_image((target_height, target_width), channels)
    processed_images += [blank_image] * (required_images - len(processed_images))

    # グリッド画像の構築
    return build_image_grid(
        processed_images,
        (grid_rows, grid_cols),
        (target_height, target_width)
    )
```

```python
import numpy as np
%config InlineBackend.print_figure_kwargs={'facecolor': "w", 'bbox_inches': 'tight', 'pad_inches': 0}

def pad_image(img, target_size):
    new_img = Image.new('RGB', target_size, color='white')
    paste_x = (target_size[0] - img.width) // 2
    paste_y = (target_size[1] - img.height) // 2
    new_img.paste(img, (paste_x, paste_y))
    return new_img

# 画像を読み込んでリサイズする関数
def load_and_resize_images(files):
    images = []
    for file in files:
        if file.lower().endswith('.png'):
            img_path = file
            img = Image.open(img_path).convert('RGB')
            # img = img.resize((img.size[0]//4,img.size[1]//4))
            img.thumbnail((1216//4,832//4))
            img = pad_image(img, (1216//4,832//4))
            images.append(np.asarray(img))
    return images

def plot_images(image_files):
    # 画像を読み込む
    images = load_and_resize_images(image_files)

    # グリッド画像を作成
    grid_image = image_grid(images)
    plt.figure(figsize=(grid_image.shape[1]//50, grid_image.shape[0]//50))
    plt.imshow(grid_image)
    plt.axis('off')
    plt.show()
```
