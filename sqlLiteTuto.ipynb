```python
import sqlite3
```

```python
conn = sqlite3.connect("test.db")
```

```python
cur = conn.cursor()
```

```python
cur.execute('''CREATE TABLE IF NOT EXISTS images(
    id INTEGER PRIMARY KEY,
    path TEXT,
    description TEXT,
    detail TEXT
)''')
```

```python
cur.execute('INSERT INTO images(path, description, detail) VALUES("mypath","mydescription","mydetail")')
```

```python
cur.execute('INSERT INTO images(path, description, detail) VALUES("2path","2description","2detail")')
```

```python
conn.commit()
```

```python
cur.close()
conn.close()
```
