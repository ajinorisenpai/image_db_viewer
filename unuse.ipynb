```python
def open_text_file(text_file, startIndex=0):
    images_with_captions = []
    with open(text_file, 'r') as f:
         # 指定された行まで読み飛ばす
        for _ in range(startIndex):
            next(f, None)
        for line in f:
            if line.strip():
                parts = line.strip().split('^^^')
                if len(parts) >= 2:
                    img_path = parts[0]
                    caption = parts[1]
                    images_with_captions.append((img_path, caption))
    return images_with_captions
```
